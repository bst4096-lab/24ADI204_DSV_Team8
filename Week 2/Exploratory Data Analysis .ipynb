# Ames Housing Dataset – Data Visualization

This notebook performs exploratory data visualization on the **Ames Housing uncleaned dataset** using **Matplotlib** and **Seaborn**.

### Visualizations included
1. Dataset overview
2. Missing-value analysis
3. Distribution of SalePrice
4. SalePrice by Overall Quality
5. Gr Liv Area vs SalePrice
6. Correlation heatmap
7. Average SalePrice by Neighborhood
8. Distribution of house construction year
9. Distribution of living area
10. Boxplot of SalePrice by House Style


## Loading the Dataset and Required Libraries

We begin by importing the Python libraries needed for this analysis and loading the Ames Housing dataset. Pandas is used for handling the data, while NumPy, Matplotlib, and Seaborn help us explore and visualize the dataset.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Plot settings
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)

# Load dataset
file_path = "AmesHousing_Uncleaned.xlsx"
df = pd.read_excel(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
df.head()
```

## Getting an Overview of the Dataset

Before creating individual visualizations, it is useful to understand the basic size and structure of the dataset. This section checks how many records and columns are available.


```python
# Dataset overview

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
print("Total missing cells:", df.isna().sum().sum())
print("\nData types:")
display(df.dtypes)

print("\nBasic statistics:")
display(df.describe(include="all").T.head(20))
```

## Checking for Missing Values

The dataset is described as uncleaned, so the first step is to look for missing values. This visualization helps identify which columns contain the most missing data and may need attention during preprocessing.


```python
# 1. Missing values by column

missing = df.isna().sum()
missing = missing[missing > 0].sort_values(ascending=False)

plt.figure(figsize=(12, 7))
sns.barplot(x=missing.values, y=missing.index)
plt.title("Missing Values by Column")
plt.xlabel("Number of Missing Cells")
plt.ylabel("Column")
plt.tight_layout()
plt.show()
```

## Exploring the Distribution of Sale Prices

Here we look at how house sale prices are distributed across the dataset. The visualization helps us understand the typical price range and whether there are unusually expensive properties affecting the distribution.


```python
# 2. Distribution of SalePrice

plt.figure(figsize=(10, 6))
sns.histplot(df["SalePrice"].dropna(), kde=True, bins=40)
plt.title("Distribution of SalePrice")
plt.xlabel("Sale Price")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## Comparing Sale Price with Overall Quality

House quality is expected to have an effect on its selling price. This visualization compares sale prices across the different Overall Quality ratings to see how strongly price changes with perceived house quality.


```python
# 3. SalePrice by Overall Quality

plt.figure(figsize=(11, 6))
sns.boxplot(data=df, x="Overall Qual", y="SalePrice")
plt.title("Sale Price by Overall Quality")
plt.xlabel("Overall Quality")
plt.ylabel("Sale Price")
plt.tight_layout()
plt.show()
```

## Relationship Between Living Area and Sale Price

The size of a house can be an important factor in its selling price. Here we examine the relationship between above-ground living area and SalePrice to see whether larger homes generally sell for more.


```python
# 4. Gr Liv Area vs SalePrice

plot_df = df[["Gr Liv Area", "SalePrice"]].dropna()

plt.figure(figsize=(10, 6))
sns.scatterplot(data=plot_df, x="Gr Liv Area", y="SalePrice", alpha=0.6)
plt.title("Above-Ground Living Area vs Sale Price")
plt.xlabel("Above-Ground Living Area (sq ft)")
plt.ylabel("Sale Price")
plt.tight_layout()
plt.show()
```

## Examining Correlations Between Numerical Variables

A correlation heatmap provides a quick way to compare relationships between important numerical variables. Strong positive or negative correlations can help identify variables that may be useful for understanding house prices.


```python
# 5. Correlation heatmap for important numerical variables

corr_cols = [
    "SalePrice", "Overall Qual", "Gr Liv Area", "Garage Cars",
    "Garage Area", "Total Bsmt SF", "1st Flr SF", "2nd Flr SF",
    "Year Built", "Year Remod/Add", "Full Bath", "TotRms AbvGrd"
]

corr = df[corr_cols].corr()

plt.figure(figsize=(12, 9))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Heatmap of Important Numerical Variables")
plt.tight_layout()
plt.show()
```

## Comparing Average Prices Across Neighborhoods

Location is another important factor in housing prices. This visualization compares the average SalePrice across different neighborhoods to see how much prices vary by area.


```python
# 6. Average SalePrice by Neighborhood

neighborhood_price = (
    df.groupby("Neighborhood", dropna=False)["SalePrice"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(13, 8))
sns.barplot(x=neighborhood_price.values, y=neighborhood_price.index)
plt.title("Average Sale Price by Neighborhood")
plt.xlabel("Average Sale Price")
plt.ylabel("Neighborhood")
plt.tight_layout()
plt.show()
```

## Exploring the Year Houses Were Built

The construction year can provide useful context about the housing stock. This distribution shows when the houses in the dataset were built and whether the data is concentrated around particular periods.


```python
# 7. Distribution of Year Built

plt.figure(figsize=(10, 6))
sns.histplot(df["Year Built"].dropna(), bins=30, kde=True)
plt.title("Distribution of House Construction Year")
plt.xlabel("Year Built")
plt.ylabel("Number of Houses")
plt.tight_layout()
plt.show()
```

## Exploring the Distribution of Living Area

Here we examine the distribution of above-ground living area. This helps us understand the typical size of the houses and identify whether a small number of unusually large properties are present.


```python
# 8. Distribution of Above-Ground Living Area

plt.figure(figsize=(10, 6))
sns.histplot(df["Gr Liv Area"].dropna(), bins=40, kde=True)
plt.title("Distribution of Above-Ground Living Area")
plt.xlabel("Living Area (sq ft)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

## Comparing Prices Across House Styles

Different house styles may have different market values. This visualization compares SalePrice across house styles to identify noticeable differences in their typical selling prices.


```python
# 9. SalePrice by House Style

plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x="House Style", y="SalePrice")
plt.title("Sale Price by House Style")
plt.xlabel("House Style")
plt.ylabel("Sale Price")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Key Observations to Explore

- **SalePrice** is generally right-skewed, with a smaller number of high-priced properties.
- **Overall Qual** is strongly associated with house sale prices.
- **Gr Liv Area** generally has a positive relationship with **SalePrice**.
- The correlation heatmap helps identify numerical variables that may be useful for house-price prediction.
- Average house prices vary considerably between **Neighborhoods**.
- Construction year and living area provide useful information about the characteristics of the properties.
- Missing values should be handled carefully before further statistical analysis or machine-learning modeling.
